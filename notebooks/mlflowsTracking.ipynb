```python
import mlflow
import mlflow.sklearn
import os
import pickle
```

```python

# Set MLflow Tracking URI (Can be a local or remote server)
mlflow.set_tracking_uri("mlruns/")

# Define Experiment
mlflow.set_experiment("Tesla_Stock_Forecasting")

with mlflow.start_run():
    # Log Model Parameters
    model_params = {"order": (2, 1, 2), "seasonal_order": (1, 1, 1, 12)}
    mlflow.log_params(model_params)

   
```

```python
 # Load SARIMAX Model
model_path = "../model/sarimax_model.pkl"
with open(model_path, "rb") as file:
        sarimax_model = pickle.load(file)

    # Log Model File
mlflow.sklearn.log_model(sarimax_model, "sarimax_model")

    # Log Metrics (Example)
metrics = {"MAE": 1.23, "RMSE": 2.34}  # Replace with actual values
mlflow.log_metrics(metrics)

print("Model Tracking Completed.")
```
